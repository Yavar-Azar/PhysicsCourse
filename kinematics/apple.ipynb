```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display, clear_output
import ipywidgets as widgets

# Physics constants
g = 9.81
x_target = 20
y_target = 10
collision_threshold = 0.1

# Simulate projectile and falling object
def simulate(v0):
    angle = np.arctan2(y_target, x_target)
    vx, vy = v0 * np.cos(angle), v0 * np.sin(angle)
    t = np.linspace(0, 2 * y_target / g, 500)
    x_proj = vx * t
    y_proj = vy * t - 0.5 * g * t**2
    x_tgt = np.full_like(t, x_target)
    y_tgt = y_target - 0.5 * g * t**2
    return t, x_proj, y_proj, x_tgt, y_tgt, vx, vy

# Animation function
def run_animation(v0):
    t, x_b, y_b, x_a, y_a, vx, vy = simulate(v0)

    plt.style.use('dark_background')
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(0, 25)
    ax.set_ylim(0, 15)
    ax.set_aspect('equal')
    ax.set_title("🎯 Bullet vs Falling Apple", fontsize=16, color='#00f5d4')
    ax.set_xlabel("X (m)", color='white')
    ax.set_ylabel("Y (m)", color='white')
    ax.tick_params(colors='white')

    # Colors and styles
    bullet_color = '#f4a261'
    apple_color = '#2ec4b6'
    trail_color_b = '#ffbe0b'
    trail_color_a = '#cbf3f0'

    # Initial velocity vector
    ax.arrow(0, 0, vx * 0.1, vy * 0.1, head_width=0.5, head_length=0.5,
             color='#00f5d4', label='Velocity Vector')

    # Initial apple position marker (horizontal line)
    ax.hlines(y_target, x_target - 0.5, x_target + 0.5, colors='white', linestyles='dotted', linewidth=1.5)

    # Trail placeholders
    trail_b, = ax.plot([], [], linestyle='--', color=trail_color_b, linewidth=1.5)
    trail_a, = ax.plot([], [], linestyle='--', color=trail_color_a, linewidth=1.5)

    # Unicode emoji text elements (WORKING in Colab)
    bullet_txt = ax.text(0, 0, 	u"\u26AB", fontsize=16, ha='center', va='center')
    apple_txt = ax.text(0, 0, "\u26AB", color='red', fontsize=18, ha='center', va='center')

    # Text annotations (right of apple)
    text_dist = ax.text(x_target + 1.0, 12.5, "", fontsize=11, color='cyan',
                        bbox=dict(facecolor='black', alpha=0.8, edgecolor='cyan'))
    text_fall = ax.text(x_target + 1.0, 11.3, "", fontsize=11, color='lime',
                        bbox=dict(facecolor='black', alpha=0.8, edgecolor='lime'))

    ax.legend(loc='lower left', facecolor='black', edgecolor='white', labelcolor='white')

    xb_data, yb_data = [], []
    ya_data = []
    collided = [False]

    def init():
        bullet_txt.set_position((0, 0))
        apple_txt.set_position((x_target, y_target))
        trail_b.set_data([], [])
        trail_a.set_data([], [])
        return bullet_txt, apple_txt, trail_b, trail_a, text_dist, text_fall

    def update(i):
        if collided[0]:
            return bullet_txt, apple_txt, trail_b, trail_a, text_dist, text_fall

        xb, yb = x_b[i], y_b[i]
        xa, ya = x_a[i], y_a[i]
        dist = np.hypot(xb - xa, yb - ya)
        fall = y_target - ya

        if dist < collision_threshold:
            collided[0] = True
            ax.text(xa, ya, u"\u26A1", color='yellow', fontsize=28, ha='center', va='center')
            return bullet_txt, apple_txt, trail_b, trail_a, text_dist, text_fall

        xb_data.append(xb)
        yb_data.append(yb)
        ya_data.append(ya)

        bullet_txt.set_position((xb, yb))
        apple_txt.set_position((xa, ya))
        trail_b.set_data(xb_data, yb_data)
        trail_a.set_data(x_a[:i+1], ya_data)
        text_dist.set_text(f"📏 Distance: {dist:.2f} m")
        text_fall.set_text(f"⬇️ Fall: {fall:.2f} m")

        return bullet_txt, apple_txt, trail_b, trail_a, text_dist, text_fall

    ani = FuncAnimation(fig, update, frames=len(t), init_func=init, blit=True, interval=20)
    plt.close()
    display(HTML(ani.to_jshtml()))

# Widgets
velocity_slider = widgets.FloatSlider(value=25, min=5, max=50, step=1, description='Bullet Speed:')
run_button = widgets.Button(description='Run Simulation')

def on_click(b):
    clear_output(wait=True)
    display(ui)
    run_animation(velocity_slider.value)

run_button.on_click(on_click)
ui = widgets.VBox([velocity_slider, run_button])
display(ui)
```
